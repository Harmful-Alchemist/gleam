```python
mapping = {"Deno": rbtree, "SpiderMonkey": rbtree_spidermonkey, "Bun": rbtree_bun}

for name in ["Deno", "SpiderMonkey", "Bun"]:
    rbtree = mapping[name]
    fig, ax = plt.subplots()
    
    names = list(map(lambda m: m.name, rbtree))
    avgs = list(map(lambda m: m.avg, rbtree))
    errsMin = list(map(lambda m: m.avg-m.minn,rbtree))
    errsMax = list(map(lambda m: m.maxx-m.avg,rbtree))
    errs = [errsMin,errsMax]
    
    bar_colors = ['tab:red', 'tab:blue', 'tab:purple', 'tab:orange', 'tab:pink']
    
    ax.bar(names, avgs, color=bar_colors, yerr=errs, capsize=5.0)
    
    ax.set_ylabel('microseconds')
    
    ax.set_xticklabels(ax.get_xticklabels(), rotation=40, ha="right")
    ax.set(ylim=[100,None])
    
    plt.tight_layout()
    plt.title(name)
    plt.savefig(f'rbtreeperf_{name}.svg')
    plt.show()
```
